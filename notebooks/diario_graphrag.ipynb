```python
import os
from pathlib import Path
from internal_graphrag.graph import Graph

uri = os.getenv("NEO4J_URI", "bolt://localhost:17687")
user = os.getenv("NEO4J_USER", "neo4j")
password = os.environ["NEO4J_PASSWORD"]
graph = Graph(uri, user, password)
```

## Source-backed passage search

Each result includes the official publication URL and the exact indexed passage.


```python
results = graph.search("IVA construção", limit=10)
for row in results:
    print(row["title"], row["publication_date"], row["url"])
    print(row["passage"][:500], "\n")
```
